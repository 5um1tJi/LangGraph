```python
from langgraph.graph import StateGraph
from typing import TypedDict
from langgraph.graph import START, END
```

```python
class bmistate(TypedDict):
    hei: float
    wei: float
    bmi: float 
    lab: str
```

```python
def bmicalc(ele : bmistate) -> bmistate:
    hei = ele["hei"]
    wei = ele["wei"]
    bmi = round(wei/(hei**2), 2)
    
    ele["bmi"] = bmi
    return ele
```

```python
def label(state: bmistate) -> bmistate:
    bmi = state["bmi"]
    if(bmi < 18.5):
        state["lab"] = "Patla"
    elif(bmi >= 18.5 and bmi <= 24.9):
        state["lab"] = "Theek-Thak"
    elif(bmi >= 25 and bmi < 29.9):
        state["lab"] = "Mota"
    else: state["lab"] = "Genda"
    return state
```

```python
graph = StateGraph(bmistate)
graph.add_node("bmicalc", bmicalc)
graph.add_node("label", label)
graph.add_edge(START, "bmicalc")
graph.add_edge("bmicalc", "label")
graph.add_edge("label", END)

kuchbhi = graph.compile()
```

```python
shuruka = {"hei" : 1.66, "wei": 57}
output = kuchbhi.invoke(shuruka)
print(output)
```
