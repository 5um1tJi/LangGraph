```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=api
)
```

```python
class llmstate(TypedDict):
    que: str
    ans: str
```

```python
def call(ele : llmstate) -> llmstate:
    que = ele["que"]
    prompt = f"Answer the following question {que}"
    answer = llm.invoke(prompt)
    ele["ans"] = answer.content
    return ele
```

```python
graph = StateGraph(llmstate)

graph.add_node("call", call)

graph.add_edge(START, "call")
graph.add_edge("call", END)

comp = graph.compile()
```

```python
input = {"que": "who is jon jones, why people call him the goat"}
output = comp.invoke(input)
print(output["ans"])
```
